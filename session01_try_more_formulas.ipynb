# Harmonic mean
average to find central tendency of rates or ratios \
\
$N / \sum_{i=1}^{N} \frac{1}{x_i}$

```python
def harmonic_mean(numbers):
  sum_reciprosal = 0
  total_number = len(numbers)

  for value in numbers:
    reciprosal = 1 / value
    sum_reciprosal = sum_reciprosal + reciprosal

  mean = (total_number) / (sum_reciprosal)
  return mean
```

```python
import numpy as np
```

```python
def harmonic_mean_numpy(numbers):
  mean = len(numbers) / sum(1/np.array(numbers))
  return mean
```

```python
numbers = [2, 4, 8, 16, 32]

print(f"mean calculated by hand: {harmonic_mean(numbers)}")
print(f"mean calculated with numpy: {harmonic_mean_numpy(numbers)}")
```

# GC content

fraction of an obtained sequence containing of G or C \
\
$\dfrac{1}{L}\sum_{i=1}^{L} \mathbb{1}[\,b_i \in \{G, C\}\,]$

```python
def gc_content(sequence):
  sum_GC = 0
  length = 0
  for nucleotide in sequence:
    if nucleotide == "C" or nucleotide == "G":
      sum_GC = sum_GC + 1
    length = length + 1

  GC_content = 1 / length * sum_GC
  return GC_content
```

```python
import numpy as np

def GC_numpy(sequence):
  sequence_array = np.array(list(sequence))
  GC_content = (1 / len(sequence)) * np.sum((sequence_array == "C")|(sequence_array == "G"))
  return GC_content
```

```python
seq = "AGCTACGTCTAG"
print(f"The GC-content of the sequence {seq} is: {gc_content(seq)}.")

print(f"The GC-content of the sequence {seq} calculated using the aid of numpy is: {GC_numpy(seq)}")
```
